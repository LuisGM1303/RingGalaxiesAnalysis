Los datasets contienen información tabular asociada a galaxias observadas. Los campos que conforman estos datasets son:

| Columna | Descripción |
|---|---|
| objID / name | Identificador único de la galaxia |

| dec | Declinación astronómica |

| anillos / label | Etiqueta asociada a la presencia o tipo de anillo |

**IMPORTANTE:**

Los datasets actúan como catálogos de referencia astronómica cuyo propósito principal es identificar y localizar objetos (galaxias en este este ejercicio) específicos en el espacio mediante coordenadas celestes (`ra`, `dec`). Aunque el objetivo final del proyecto es entrenar modelos de aprendizaje automático sobre imágenes astronómicas, el conjunto de datos inicial no contiene directamente estas imágenes, sino la información necesaria para recuperarlas.

Estas coordenadas serán utilizadas para descargar las imágenes FITS correspondientes desde el servicio Legacy Survey (https://www.legacysurvey.org), que proporciona cutouts fotométricos de alta resolución en múltiples bandas espectrales (g, r, z). De esta manera, el catálogo tabular actúa como un índice de referencia que vincula cada galaxia con su representación visual, permitiendo así construir un dataset de imágenes etiquetadas apto para tareas de clasificación supervisada.

## Explicación de los Campos del Catálogo de Galaxias

LOs campos objID/name, ra, dec y z son las "etiquetas" que nos ayudan a identificar y localizar cada galaxia en el "universo". Cada galaxia es un objeto en un mapa del cielo, y estos campos nos dan su "dirección" y algunas propiedades básicas. A continuación, explicamos con detalle el significado de estos campos:

**objID (Object Identifier)**

  - **¿Qué es?** Es un identificador único, como un número de serie o un código de barras, asignado a cada galaxia en el catálogo. No se repite para ninguna otra galaxia, lo que lo hace perfecto para referenciarla sin confusiones.
  - **¿Por qué es importante?** Nos permite rastrear galaxias específicas a lo largo de diferentes análisis o datasets. Por ejemplo, si descargamos imágenes de una galaxia, podemos usar su objID para relacionarla con sus datos tabulares.
  - **Ejemplo:** En el dataset, ves números largos como `1237648721210769659`. Estos vienen del Sloan Digital Sky Survey (SDSS), que es una gran base de datos astronómica.
  - **Fuentes para más info:**
    - [Página oficial del SDSS](https://www.sdss.org/) – Explica cómo se generan estos IDs.
    - [Wikipedia: Sloan Digital Sky Survey](https://en.wikipedia.org/wiki/Sloan_Digital_Sky_Survey) – Una introducción amigable al proyecto.

**ra (Right Ascension)**

  - **¿Qué es?** La Ascensión Recta (RA) es como la "longitud" en el mapa del cielo. Es una coordenada angular que mide la posición este-oeste de una galaxia en la esfera celeste, expresada en grados, horas o grados decimales.
  - **¿Por qué es importante?** Junto con la declinación, nos da la ubicación exacta de la galaxia. Es esencial para apuntar telescopios o descargar imágenes desde servicios como Legacy Survey, ya que especifica dónde buscar en el cielo.
  - **Un poco más técnico (pero amigable):** Se mide desde el equinoccio vernal (un punto de referencia en el cielo) y va de 0° a 360°. En astronomía, a veces se expresa en horas (0h a 24h), donde 1 hora = 15 grados.
  - **Ejemplo:** En el dataset, valores como `134.44717` grados. Esto nos dice que la galaxia está en una posición específica, como si fuera una dirección GPS en el cielo.
  - **Fuentes para más info:**
    - [Wikipedia: Ascensión Recta](https://en.wikipedia.org/wiki/Right_ascension) – Definición clara con ejemplos.
    - [Astronomía básica en Khan Academy](https://www.khanacademy.org/science/physics/special-relativity) – Videos introductorios sobre coordenadas celestes (relacionado con relatividad, pero útil).

**dec (Declination)**

  - **¿Qué es?** La Declinación (Dec) es como la "latitud" en el mapa del cielo. Es otra coordenada angular que mide la posición norte-sur de una galaxia, desde -90° (polo sur celeste) hasta +90° (polo norte celeste).
  - **¿Por qué es importante?** Completa la ubicación junto con RA. Sin estas dos, no podríamos "apuntar" a la galaxia para observarla o descargar sus imágenes. Es clave para la astronomía moderna, ya que permite localizar objetos con precisión.
  - **Un poco más técnico (pero amigable):** Se mide desde el ecuador celeste. Valores positivos son hacia el norte, negativos hacia el sur. Por ejemplo, el ecuador celeste pasa por el ecuador terrestre.
  - **Ejemplo:** En el dataset, valores como `-0.199973` grados, indicando una posición ligeramente al sur del ecuador celeste.
  - **Fuentes para más info:**
    - [Wikipedia: Declinación (astronomía)](https://en.wikipedia.org/wiki/Declination) – Explicación detallada.
    - [Sitio de la NASA: Coordenadas celestes](https://science.nasa.gov/missions/hubble/coordinates) – Guía visual y sencilla.

**z (Redshift)**

  - **¿Qué es?** El corrimiento al rojo (redshift) es una medida de cuánto se ha "estirado" la luz de la galaxia debido al movimiento o la expansión del universo. Se representa como `z`, y valores mayores indican que la galaxia se aleja más rápido de nosotros.
  - **¿Por qué es importante?** Nos da una idea de la distancia y la velocidad de la galaxia. En cosmología, es fundamental para estudiar la evolución del universo. Para nuestro proyecto, ayuda a filtrar galaxias en rangos de distancia similares, asegurando datos consistentes.
  - **Un poco más técnico (pero amigable):** Si `z = 0`, la galaxia está en reposo relativo a nosotros. Para `z > 0`, la luz se desplaza hacia el rojo (longitudes de onda más largas). Se calcula con la fórmula `(λ_observada - λ_emitida) / λ_emitida`, donde λ es la longitud de onda.
  - **Ejemplo:** En el dataset, valores como `0.028206`, lo que sugiere una galaxia no demasiado lejana, con un ligero corrimiento.
  - **Fuentes para más info:**
    - [Wikipedia: Corrimiento al rojo](https://en.wikipedia.org/wiki/Redshift) – Explicación completa con animaciones.
    - [Cosmología en el Instituto de Astronomía UNAM](https://www.astro.unam.mx/~flores/cosmologia/) – Recursos en español sobre redshift y expansión del universo.


¿Cómo se usan juntos estos campos?

Estos cuatro campos forman la "identidad" básica de cada galaxia en el catálogo. Con `objID`, `ra`, `dec` y `z`, podemos:

- Identificarla de forma única.
- Localizarla en el cielo para descargar imágenes FITS desde Legacy Survey.
- Analizar su distancia y movimiento para estudios morfológicos, como detectar anillos.

## Interpretación de etiquetas de anillos

Las etiquetas asociadas a estructuras de anillos siguen la siguiente codificación:

| Valor | Tipo de estructura |
|---|---|
| 0 | Sin anillo |
| 2 | Anillo nuclear |
| 4 | Anillo interno |
| 8 | Anillo externo |
| 12 | Anillo interno + externo |
| 16 | Pseudoanillo |

Para las primeras etapas del proyecto, el problema se simplifica a una tarea de clasificación binaria:

- 0 → galaxias sin anillo
- 1 → galaxias con anillo interno

donde las galaxias con etiquetas originales 4 y 12 son consideradas positivas para la clase de interés.

## Relación con las imágenes astronómicas

Recordemos que estos datasets no contienen directamente las imágenes astronómicas, sino la información necesaria para recuperarlas desde el servicio Legacy Survey mediante coordenadas astronómicas (`ra`, `dec`).

Las imágenes son descargadas en formato `.fits`, y contienen información multibanda correspondiente a las bandas espectrales:

- g
- r
- z

Estas imágenes constituyen la entrada principal para las etapas posteriores de procesamiento y modelado.

# Data Processing

## Procesamiento y consolidación de catálogos astronómicos

Una vez cargados ambos catálogos, se realiza una etapa de procesamiento cuyo objetivo es construir un único dataset cuya información nos ayudará a plantear un problema  de clasificación binaria en el que entrenaremos un modelo para etiquetar si una galaxia cuenta con anillo interno o no.

El objetivo principal de este procesamiento es transformar los catálogos originales en un conjunto de datos homogéneo que nos permitirá proceder con la descarga de las imágenes astronómicas asociadas a cada objeto del dataset.

Los pasos asociados a este procesamiento son:

**1. Filtrado por estructuras de interés**

Dado que el problema inicial del proyecto se enfoca en la detección de **anillos internos**, se seleccionan únicamente galaxias pertenecientes a las siguientes categorías morfológicas:

| Código | Descripción |
|---|---|
| 0 | Sin anillo |
| 4 | Anillo interno |
| 12 | Anillo interno + anillo externo |

Las demás categorías (`2`, `8`, `16`) quedan fuera del alcance de esta primera etapa.

**2. Estandarización de columnas**

Los catálogos originales presentan diferencias en nomenclatura y estructura, por lo que se realiza una homologación de columnas para garantizar consistencia entre ambas fuentes.

Se adopta el siguiente esquema común:

| Columna | Descripción |
|---|---|
| `name` | Identificador único de la galaxia |
| `objra` | Ascensión recta |
| `objdec` | Declinación |
| `nsa_z` | Corrimiento al rojo |
| `anillos` | Etiqueta original de morfología |
| `source` | Catálogo de origen |

Esto permite trabajar ambos datasets como una sola fuente integrada.

**3. Integración de catálogos**

Posteriormente, ambos subconjuntos filtrados son combinados en un único dataframe.

Se agrega una columna `source` para conservar trazabilidad sobre el origen de cada observación:

- `sdss`: galaxias provenientes del catálogo basado en Fernández et al.
- `manga`: galaxias provenientes del catálogo complementario MaNGA generado por los investigadores.

**4. Generación de etiqueta binaria**

Para simplificar el problema hacia una primera tarea de clasificación binaria, se define una nueva variable objetivo:

- `label = 0` → galaxias sin anillo
- `label = 1` → galaxias con anillo interno

En esta transformación, tanto las galaxias etiquetadas originalmente como `4` (anillo interno) como `12` (anillo interno + externo) son agrupadas en la clase positiva.

**5. Persistencia del dataset procesado**

Finalmente, el dataset consolidado se guarda como:

```text
data/raw/catalogs/classification_dataset.csv
```

Este archivo constituye la base principal para las siguientes etapas del proyecto:

- descarga de imágenes desde Legacy Survey,
- análisis exploratorio visual,
- procesamiento de imágenes `.fits`,

````python
# ============================================================
# 1. Filtrado por anillos de interés
# ============================================================

# Anillos seleccionados:
# 0  -> sin anillo
# 4  -> anillo interno
# 12 -> anillo interno + externo
RINGS = [0, 4, 12]

# --------------------------
# Dataset Fernández et al.
# --------------------------

sdss_filtrado = fernandez_df[
    fernandez_df["anillos"].isin(RINGS)
].copy()

sdss_filtrado = sdss_filtrado.rename(
    columns={
        "ra": "objra",
        "dec": "objdec",
        "z": "nsa_z",
    }
)

sdss_filtrado["source"] = "sdss"
sdss_filtrado["name"] = sdss_filtrado["objID"].astype(str)

sdss_filtrado = sdss_filtrado[
    ["name", "objra", "objdec", "nsa_z", "anillos", "source"]
]

print(f"SDSS filtrado: {len(sdss_filtrado)} galaxias")
print(sdss_filtrado["anillos"].value_counts().sort_index())


# --------------------------
# Dataset MaNGA
# --------------------------

manga_filtrado = manga_rings_df[
    manga_rings_df["anillos"].isin(RINGS)
].copy()

manga_filtrado["source"] = "manga"

manga_filtrado = manga_filtrado[
    ["name", "objra", "objdec", "nsa_z", "anillos", "source"]
]

print(f"\nMaNGA filtrado: {len(manga_filtrado)} galaxias")
print(manga_filtrado["anillos"].value_counts().sort_index())
````

````python
# ============================================================
# 2. Combinación y etiquetado binario
# ============================================================

df = pd.concat(
    [sdss_filtrado, manga_filtrado],
    ignore_index=True
)

# Label binario:
# 0 -> sin anillo
# 1 -> con anillo interno (4 o 12)

df["label"] = df["anillos"].apply(
    lambda x: 0 if x == 0 else 1
)

print("\nDataset combinado:")
print(
    df["label"].value_counts().rename(
        {
            0: "Sin anillo (0)",
            1: "Anillo interno (1)",
        }
    )
)

print(f"\nTotal: {len(df)} galaxias")

display(df.head())
````

````python
# ============================================================
# 3. Distribución de clases
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 4)
)

# --------------------------
# Distribución binaria
# --------------------------

counts = df["label"].value_counts().sort_index()

axes[0].bar(
    ["Sin anillo\n(label=0)", "Anillo interno\n(label=1)"],
    counts.values
)

axes[0].set_title("Distribución de clases")
axes[0].set_ylabel("Cantidad de galaxias")

for i, v in enumerate(counts.values):
    axes[0].text(
        i,
        v + 10,
        str(v),
        ha="center",
        fontweight="bold"
    )


# --------------------------
# Distribución por anillo original
# --------------------------

ring_counts = df["anillos"].value_counts().sort_index()

labels_map = {
    0: "Sin anillo\n(0)",
    4: "Interno\n(4)",
    12: "Interno+Externo\n(12)",
}

axes[1].bar(
    [labels_map[k] for k in ring_counts.index],
    ring_counts.values
)

axes[1].set_title("Detalle por tipo de anillo")
axes[1].set_ylabel("Cantidad de galaxias")

for i, v in enumerate(ring_counts.values):
    axes[1].text(
        i,
        v + 10,
        str(v),
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()

plt.savefig(
    "../data/raw/catalogs/class_distribution.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
````

````python
# ============================================================
# 4. Persistencia del dataset final
# ============================================================

output_path = "../data/raw/catalogs/classification_dataset.csv"

df.to_csv(
    output_path,
    index=False
)

print(f"\nDataset guardado en: {output_path}")
print(f"Columnas: {list(df.columns)}")

print("\nResumen final:")
print(
    df.groupby(
        ["source", "label"]
    )["name"].count().rename("galaxias")
)
````

# Data Collection

**Descarga concurrente de imágenes astronómicas**

A medida que el volumen de galaxias aumenta, la descarga secuencial de imágenes FITS se vuelve ineficiente debido a:

- latencia de red,
- tiempo de respuesta del servidor,
- y tamaño de los archivos astronómicos.

Para mejorar el rendimiento del pipeline de adquisición de datos, se implementa un sistema de descarga asíncrona basado en:

- `asyncio`
- `aiohttp`

**Objetivos de la estrategia concurrente**

La implementación concurrente permite:

- descargar múltiples galaxias simultáneamente,
- aprovechar tiempos de espera de red,
- reducir significativamente el tiempo total de adquisición,
- controlar el número máximo de requests concurrentes,
- manejar reintentos automáticos ante fallos temporales.

**Descarga por batches**

El pipeline utiliza batches concurrentes para evitar:

- saturar el servidor Legacy Survey,
- bloqueos por exceso de requests,
- consumo excesivo de memoria,
- y problemas de estabilidad de red.

Cada batch procesa un número controlado de galaxias simultáneamente.

**Beneficios para el proyecto**

La descarga concurrente es especialmente importante debido a que:

- cada galaxia requiere múltiples imágenes FITS (`g`, `r`, `z`),
- el dataset crecerá progresivamente,
- futuros experimentos requerirán redescargas parciales,
- y el proyecto potencialmente escalará hacia datasets astronómicos masivos.

Esta estrategia mejora considerablemente la escalabilidad del pipeline CRISP-ML.

````python
# ============================================================
# Configuración general
# ============================================================

PIXSCALE = 0.262
SIZE = 800

LAYER = "ls-dr10"

BANDS = ["g", "r", "z"]

# Concurrencia máxima
MAX_CONCURRENT_REQUESTS = 10

# Reintentos automáticos
MAX_RETRIES = 3

# Timeout por request
REQUEST_TIMEOUT = 60

# Tiempo de espera entre reintentos (en segundos)
RETRY_DELAY = 120
````

````python
# ============================================================
# Paths
# ============================================================

DATASET_PATH = Path(
    "../data/raw/catalogs/classification_dataset.csv"
)

BASE_OUTPUT_DIR = Path(
    "../data/raw/legacy_fits_files"
)

# ============================================================
# Directorios por banda y label
# ============================================================

LABELS = [0, 1]

BAND_DIRS = {
    label: {
        band: BASE_OUTPUT_DIR / f"label_{label}" / band
        for band in BANDS
    }
    for label in LABELS
}

for band_dirs in BAND_DIRS.values():
    for directory in band_dirs.values():
        directory.mkdir(
            parents=True,
            exist_ok=True
        )

def get_band_dir(band: str, label: int) -> Path:
    return BAND_DIRS[label][band]

# ============================================================
# Dataset
# ============================================================

df = pd.read_csv(DATASET_PATH)

print(f"Galaxias totales: {len(df)}")

# ============================================================
# URL Builder
# ============================================================

BASE_URL = (
    "http://legacysurvey.org/viewer/fits-cutout"
)

def build_url(
    ra: float,
    dec: float,
    band: str
) -> str:

    return (
        f"{BASE_URL}"
        f"?ra={ra:.6f}"
        f"&dec={dec:.6f}"
        f"&size={SIZE}"
        f"&layer={LAYER}"
        f"&pixscale={PIXSCALE}"
        f"&bands={band}"
    )
````

````python
# ============================================================
# Utilidades
# ============================================================

def safe_name(name: str) -> str:

    return (
        name
        .replace("/", "_")
        .replace(" ", "_")
        .replace(":", "_")
    )
````

````python
# ============================================================
# Descarga individual asíncrona
# ============================================================

async def download_band(
    session: aiohttp.ClientSession,
    semaphore: asyncio.Semaphore,
    galaxy_name: str,
    ra: float,
    dec: float,
    band: str,
    label: int
):
    """
    Descarga una banda FITS específica
    de manera asíncrona.
    """

    output_dir = BAND_DIRS[label][band]

    output_path = output_dir / f"{galaxy_name}_{band}.fits"

    if output_path.exists():
        print(
            f"[info] {galaxy_name} - {band} - Archivo ya existe. Saltando descarga."
        )
        return {
            "name": galaxy_name,
            "band": band,
            "status": "skip",
            "path": str(output_path)
        }

    url = build_url(ra, dec, band)
    timeout = aiohttp.ClientTimeout(total=REQUEST_TIMEOUT*10)

    async with semaphore:
        for attempt in range(1, MAX_RETRIES + 1):
            try:
                async with session.get(url, timeout=timeout) as response:
                    if response.status == 429:
                        retry_after = response.headers.get("Retry-After")
                        try:
                            wait_time = int(float(retry_after))
                        except (TypeError, ValueError):
                            wait_time = RETRY_DELAY * attempt

                        print(
                            f"[warning] {galaxy_name} - {band} - 429 Too Many Requests "
                            f"(intent {attempt}/{MAX_RETRIES}). Esperando {wait_time}s..."
                        )
                        await asyncio.sleep(wait_time)
                        continue

                    response.raise_for_status()

                    content = await response.read()

                    if content[:6] != b"SIMPLE":
                        print(
                            f"[error] {galaxy_name} - {band} - Contenido inválido."
                        )
                        return {
                            "name": galaxy_name,
                            "band": band,
                            "status": "invalid_fits",
                            "path": None
                        }

                    print(
                        f"[success] {galaxy_name} - {band} - Descarga exitosa."
                    )
                    
                    output_path.write_bytes(content)

                    return {
                        "name": galaxy_name,
                        "band": band,
                        "status": "ok",
                        "path": str(output_path)
                    }

            except aiohttp.ClientResponseError as e:
                if e.status == 429:
                    wait_time = RETRY_DELAY * attempt * 10
                    print(
                        f"[warning] {galaxy_name} - {band} - 429 Too Many Requests "
                        f"(ClientResponseError, intent {attempt}/{MAX_RETRIES}). Esperando {wait_time}s..."
                    )
                    await asyncio.sleep(wait_time)
                    continue

                print(
                    f"[error] {galaxy_name} - {band} - "
                    f"Attempt {attempt}/{MAX_RETRIES} - {e.status} {e.message}"
                )

            except (aiohttp.ServerTimeoutError, asyncio.TimeoutError) as e:
                print(
                    f"[error] {galaxy_name} - {band} - "
                    f"Timeout en attempt {attempt}/{MAX_RETRIES}: {e}"
                )

            except Exception as e:
                print(
                    f"[error] {galaxy_name} - {band} - "
                    f"Attempt {attempt}/{MAX_RETRIES} - {str(e)}"
                )

            if attempt == MAX_RETRIES:
                return {
                    "name": galaxy_name,
                    "band": band,
                    "status": f"error: {str(e)}",
                    "path": None
                }

            await asyncio.sleep(RETRY_DELAY)
````

````python
# ============================================================
# Descarga concurrente de galaxia
# ============================================================

async def download_galaxy(
    session: aiohttp.ClientSession,
    semaphore: asyncio.Semaphore,
    row: pd.Series
):
    """
    Descarga concurrentemente las bandas
    g, r, z de una galaxia.
    """

    galaxy_name = safe_name(
        str(row["name"])
    )

    ra = float(row["objra"])
    dec = float(row["objdec"])
    label = int(row["label"])

    tasks = []

    for band in BANDS:

        task = download_band(
            session=session,
            semaphore=semaphore,
            galaxy_name=galaxy_name,
            ra=ra,
            dec=dec,
            band=band,
            label=label
        )

        tasks.append(task)

    results = await asyncio.gather(*tasks)

    print(f"[done] {galaxy_name}")

    return results
````

````python
# ============================================================
# Batch downloader
# ============================================================

async def download_batch(
    dataframe: pd.DataFrame,
    batch_size: int = 5
):
    """
    Descarga galaxias concurrentemente
    en batches controlados.
    """

    semaphore = asyncio.Semaphore(
        MAX_CONCURRENT_REQUESTS
    )

    connector = aiohttp.TCPConnector(
        limit=MAX_CONCURRENT_REQUESTS
    )

    async with aiohttp.ClientSession(
        connector=connector
    ) as session:

        all_results = []

        for start_idx in range(
            0,
            len(dataframe),
            batch_size
        ):

            end_idx = start_idx + batch_size

            batch_df = dataframe.iloc[
                start_idx:end_idx
            ]

            print(
                f"\nProcesando batch "
                f"{start_idx}:{end_idx}"
            )

            tasks = []

            for _, row in batch_df.iterrows():

                task = download_galaxy(
                    session=session,
                    semaphore=semaphore,
                    row=row
                )

                tasks.append(task)

            batch_results = await asyncio.gather(
                *tasks,
                return_exceptions=True
            )

            for galaxy_results in batch_results:

                if isinstance(galaxy_results, Exception):
                    print(
                        "[error] Falló la descarga de una galaxia:",
                        galaxy_results
                    )
                    continue

                if galaxy_results is None:
                    print(
                        "[warning] Resultados de descarga de galaxia nulos. Se omite."
                    )
                    continue

                all_results.extend(
                    galaxy_results
                )

        return pd.DataFrame(all_results)
````

````python
def validate_dataset(
    df: pd.DataFrame,
    destination_directory: str,
    bands=("g", "r", "z")
):
    """
    Valida que todas las galaxias del dataframe
    existan correctamente en el directorio destino.

    Verifica:
    - existencia de cada banda
    - cantidad de bandas encontradas
    - galaxias completas/incompletas

    Retorna:
    - report_df
    - summary
    """

    destination_directory = Path(destination_directory)

    report = []

    for _, row in df.iterrows():

        name = safe_name(str(row["name"]))
        label = int(row["label"])

        galaxy_report = {
            "name": name,
            "label": label
        }

        found_bands = 0

        for band in bands:

            expected_path = (
                destination_directory /
                f"label_{label}" /
                band /
                f"{name}_{band}.fits"
            )

            exists = expected_path.exists()

            galaxy_report[f"{band}_exists"] = exists
            galaxy_report[f"{band}_path"] = str(expected_path)

            if exists:
                found_bands += 1

        galaxy_report["found_bands"] = found_bands
        galaxy_report["complete"] = (
            found_bands == len(bands)
        )

        report.append(galaxy_report)

    report_df = pd.DataFrame(report)

    # ========================================================
    # Resumen general
    # ========================================================

    total_galaxies = len(report_df)

    complete_galaxies = (
        report_df["complete"]
        .sum()
    )

    incomplete_galaxies = (
        total_galaxies - complete_galaxies
    )

    summary = {
        "total_galaxies": total_galaxies,
        "complete_galaxies": int(complete_galaxies),
        "incomplete_galaxies": int(incomplete_galaxies),
        "completion_rate": round(
            complete_galaxies / total_galaxies * 100,
            2
        )
    }

    return report_df, summary
````

# Data Cleaning

## Validación y limpieza de imágenes FITS

Una vez completada la descarga de imágenes astronómicas desde Legacy Survey, es necesario realizar una etapa de validación y limpieza de datos antes de comenzar el análisis exploratorio visual y el entrenamiento de modelos de aprendizaje automático.

En proyectos de visión computacional astronómica, los archivos FITS pueden presentar distintos problemas derivados de:

- errores de red durante la descarga,
- archivos corruptos,
- respuestas incompletas del servidor,
- imágenes vacías,
- valores faltantes (`NaN`),
- diferencias de resolución,
- y variaciones extremas de intensidad entre bandas espectrales.

## Objetivos de esta etapa

La limpieza de datos busca garantizar que únicamente imágenes válidas y consistentes entren al pipeline de modelado.

Para ello se realizan las siguientes verificaciones:

| Verificación | Descripción |
|---|---|
| Existencia del archivo | Validar que el FITS exista físicamente |
| Lectura correcta | Confirmar que Astropy puede abrir el archivo |
| Integridad del arreglo | Verificar dimensiones válidas |
| Presencia de NaN | Detectar valores faltantes |
| Estadísticas básicas | Obtener mínimos, máximos y desviación estándar |
| Consistencia de resolución | Confirmar tamaño homogéneo |
| Calidad visual mínima | Detectar imágenes completamente vacías |

## Importancia dentro de CRISP-ML

La calidad de los datos visuales impacta directamente:

- el análisis exploratorio,
- la estabilidad del entrenamiento,
- la convergencia de modelos,
- y la calidad de las predicciones.

En astronomía, pequeños errores instrumentales o archivos corruptos pueden introducir sesgos importantes en modelos de aprendizaje automático.

Por ello, esta etapa constituye un paso crítico antes de cualquier procesamiento avanzado.

## Organización de imágenes

Las imágenes FITS fueron organizadas utilizando la siguiente estructura:

```text
data/raw/legacy_fits_files/
├── label_0/
│   ├── g/
│   ├── r/
│   └── z/
└── label_1/
    ├── g/
    ├── r/
    └── z/
```

donde:

- `label_0` → galaxias sin anillo
- `label_1` → galaxias con anillo interno

## Resultado esperado

Al finalizar esta etapa se obtiene:

- un reporte de integridad del dataset,
- estadísticas descriptivas básicas de imágenes,
- detección de archivos problemáticos,
- y métricas iniciales de calidad para las bandas `g`, `r`, `z`.

````python
from astropy.io import fits
````

````python
# Configuración
BASE_DIR = Path(
    "../data/raw/legacy_fits_files"
)

LABELS = ["label_0", "label_1"]
BANDS = ["g", "r", "z"]

# Recolección de archivos
fits_files = []
for label in LABELS:
    for band in BANDS:
        band_dir = BASE_DIR / label / band
        files = list(
            band_dir.glob("*.fits")
        )
        for file in files:
            fits_files.append(
                {
                    "label": label,
                    "band": band,
                    "path": file
                }
            )
fits_df = pd.DataFrame(fits_files)

print("===================================")
print("Resumen inicial")
print("===================================")

print(f"Total archivos FITS: {len(fits_df)}")

display(
    fits_df.groupby(
        ["label", "band"]
    ).size()
)
````

````python
# Validación FITS
results = []
for _, row in fits_df.iterrows():
    file_path = row["path"]
    try:
        with fits.open(file_path) as hdul:
            data = hdul[0].data
            # Validación básica
            valid_shape = (
                data is not None and
                len(data.shape) >= 2
            )
            nan_count = (
                np.isnan(data).sum()
                if data is not None
                else None
            )
            stats = {
                "min": np.nanmin(data),
                "max": np.nanmax(data),
                "mean": np.nanmean(data),
                "std": np.nanstd(data),
            }
            results.append(
                {
                    "file": file_path.name,
                    "label": row["label"],
                    "band": row["band"],
                    "shape": data.shape,
                    "valid_shape": valid_shape,
                    "nan_count": int(nan_count),
                    "min": stats["min"],
                    "max": stats["max"],
                    "mean": stats["mean"],
                    "std": stats["std"],
                    "status": "ok"
                }
            )

    except Exception as e:
        results.append(
            {
                "file": file_path.name,
                "label": row["label"],
                "band": row["band"],
                "shape": None,
                "valid_shape": False,
                "nan_count": None,
                "min": None,
                "max": None,
                "mean": None,
                "std": None,
                "status": f"error: {str(e)}"
            }
        )
````

````python
# DataFrame de calidad
quality_df = pd.DataFrame(results)
print("\n===================================")
print("Estado de archivos")
print("===================================")

display(
    quality_df["status"]
    .value_counts()
)
````

````python
# ============================================================
# Validación de shapes
# ============================================================

print("\n===================================")
print("Shapes encontrados")
print("===================================")

display(
    quality_df["shape"]
    .value_counts()
)
````

````python
# ============================================================
# Estadísticas descriptivas
# ============================================================

print("\n===================================")
print("Estadísticas descriptivas")
print("===================================")

display(
    quality_df[
        ["min", "max", "mean", "std"]
    ].describe()
)
````

````python
# ============================================================
# Archivos problemáticos
# ============================================================

problematic_files = quality_df[
    (
        quality_df["status"] != "ok"
    )
    |
    (
        quality_df["nan_count"] > 0
    )
]

print("\n===================================")
print("Archivos problemáticos")
print("===================================")

display(
    problematic_files.head()
)
````

````python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from astropy.io import fits
````

````python
# ============================================================
# Configuración
# ============================================================

BASE_DIR = Path(
    "../data/raw/legacy_fits_files"
)

LABELS = ["label_0", "label_1"]
BANDS = ["g", "r", "z"]
````

````python
# ============================================================
# Función para cargar FITS
# ============================================================

def load_fits_image(path: Path):
    with fits.open(path) as hdul:
        data = hdul[0].data.astype(np.float32)
    return data
````

````python
# ============================================================
# Recolección de archivos
# ============================================================

records = []

for label in LABELS:

    for band in BANDS:

        band_dir = BASE_DIR / label / band

        for file_path in band_dir.glob("*.fits"):

            galaxy_name = (
                file_path.stem
                .replace(f"_{band}", "")
            )

            records.append(
                {
                    "label": label,
                    "band": band,
                    "galaxy": galaxy_name,
                    "path": file_path
                }
            )

eda_df = pd.DataFrame(records)

print("===================================")
print("Dataset EDA")
print("===================================")

print(f"Total imágenes: {len(eda_df)}")

display(
    eda_df.head()
)
````

````python
# Selección balanceada por label
random.seed(6)

sample_galaxies = []

for label in sorted(eda_df["label"].unique()):
    galaxies = eda_df.loc[
        eda_df["label"] == label, "galaxy"
    ].unique().tolist()

    sample_galaxies.extend(
        random.sample(
            galaxies,
            min(2, len(galaxies))
        )
    )

print("\nGalaxias seleccionadas:")
print(sample_galaxies)
````

````python
# ============================================================
# Visualización por bandas
# ============================================================

fig, axes = plt.subplots(
    len(sample_galaxies),
    len(BANDS),
    figsize=(12, 12)
)

if len(sample_galaxies) == 1:
    axes = np.expand_dims(axes, 0)

for row_idx, galaxy in enumerate(sample_galaxies):
    galaxy_df = eda_df[
        eda_df["galaxy"] == galaxy
    ]

    galaxy_label = (
        galaxy_df["label"].iloc[0]
        if not galaxy_df.empty
        else "label_unknown"
    )

    for col_idx, band in enumerate(BANDS):
        ax = axes[row_idx, col_idx]

        band_df = galaxy_df[
            galaxy_df["band"] == band
        ]

        if band_df.empty:
            ax.text(
                0.5,
                0.5,
                f"Band {band}\nNo disponible",
                ha="center",
                va="center",
                fontsize=12
            )
            ax.set_title(f"{galaxy}\n{galaxy_label}\nBand {band}")
            ax.axis("off")
            continue

        image_path = band_df["path"].iloc[0]
        image = load_fits_image(image_path)

        finite_pixels = image[np.isfinite(image)]
        if finite_pixels.size == 0:
            ax.text(
                0.5,
                0.5,
                "Imagen inválida",
                ha="center",
                va="center",
                fontsize=12
            )
            ax.set_title(f"{galaxy}\n{galaxy_label}\nBand {band}")
            ax.axis("off")
            continue

        vmin = np.percentile(finite_pixels, 1)
        vmax = np.percentile(finite_pixels, 99)

        ax.imshow(
            image,
            origin="lower",
            vmin=vmin,
            vmax=vmax,
            cmap="gray"
        )
        ax.set_title(f"{galaxy}\n{galaxy_label}\nBand {band}")
        ax.text(
            0.03,
            0.95,
            galaxy_label,
            transform=ax.transAxes,
            fontsize=10,
            color="white",
            backgroundcolor="black",
            va="top",
            ha="left"
        )
        ax.axis("off")

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# Histogramas de intensidad
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 4)
)

for idx, band in enumerate(BANDS):

    band_images = eda_df[
        eda_df["band"] == band
    ]

    pixel_values = []

    # Muestreo pequeño para eficiencia
    sampled_paths = random.sample(
        list(band_images["path"]),
        min(20, len(band_images))
    )

    for path in sampled_paths:

        image = load_fits_image(path)

        pixels = image.flatten()

        pixels = pixels[
            np.isfinite(pixels)
        ]

        pixel_values.extend(pixels)

    pixel_values = np.array(pixel_values)

    axes[idx].hist(
        pixel_values,
        bins=100,
        log=True
    )

    axes[idx].set_title(
        f"Distribución intensidades\nBand {band}"
    )

    axes[idx].set_xlabel("Intensidad")
    axes[idx].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# Estadísticas por banda
# ============================================================

stats_records = []

for _, row in eda_df.iterrows():

    image = load_fits_image(
        row["path"]
    )

    finite_pixels = image[
        np.isfinite(image)
    ]

    stats_records.append(
        {
            "label": row["label"],
            "band": row["band"],
            "mean": np.mean(finite_pixels),
            "std": np.std(finite_pixels),
            "min": np.min(finite_pixels),
            "max": np.max(finite_pixels),
        }
    )

stats_df = pd.DataFrame(stats_records)

print("\n===================================")
print("Estadísticas por banda")
print("===================================")

display(
    stats_df.groupby("band")[
        ["mean", "std", "min", "max"]
    ].describe()
)
````

````python
# ============================================================
# Comparación entre clases
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 4)
)

for idx, metric in enumerate(["mean", "std"]):
    label_0 = stats_df[
        stats_df["label"] == "label_0"
    ][metric]

    label_1 = stats_df[
        stats_df["label"] == "label_1"
    ][metric]

    axes[idx].boxplot(
        [label_0, label_1],
        tick_labels=[
            "Sin anillo",
            "Anillo interno"
        ]
    )

    axes[idx].set_title(
        f"Comparación de {metric}"
    )

plt.tight_layout()
plt.show()
````
